

## 5. Tiền xử lý dữ liệu chuẩn ngữ nghĩa miền  & One-Hot Encoding

Trong 19 cột xuất hiện `NA` ở tập dữ liệu, có tới 14–15 cột ký hiệu `NA` được sử dụng để đại diện cho **"Không có tiện ích đó" (Absence of feature)** chứ không phải dữ liệu bị mất mát ngẫu nhiên:

1. **Bảo toàn các đặc trưng tiện ích đặc thù:**
   - Chỉ loại bỏ duy nhất cột `Id` do là mã định danh số học không có tính quy luật.

2. **Xử lý đặc trưng không có tiện ích:**
   - **Nhóm phân loại:** 15 cột mang ý nghĩa "Không có tiện ích" (`PoolQC`, `MiscFeature`, `Alley`, `Fence`, `FireplaceQu`, các cột `Bsmt*` và `Garage*`, `MasVnrType`) được gán tường minh thành chuỗi `'None'` thay vì điền bằng `mode` (tránh sai lầm biến nhà không có lò sưởi/hầm/garage thành nhà có tiện ích loại trung bình).
   - **Nhóm số học:** Các biến diện tích, số lượng phòng đi kèm (`GarageArea`, `GarageCars`, `TotalBsmtSF`, `BsmtFinSF1`, `BsmtFullBath`,...) được gán bằng `0` thay vì điền trung bình.

3. **Xử lý mất mát ngẫu nhiên thực sự:**
   - Cột `LotFrontage` (chiều dài mặt tiền đất) được điền theo giá trị trung vị (`median`) của từng phân khu dân cư (`Neighborhood`).
   - Các biến phân loại thiếu rải rác còn lại được điền bằng `mode`; các biến số còn lại được điền bằng `median`.

4. **Mã hóa đặc trưng & Chuẩn hóa nhãn:**
   - Chuyển `MSSubClass` về kiểu chuỗi vì bản chất là mã phân loại kiến trúc.
   - Thực hiện One-Hot Encoding với `pd.get_dummies(..., drop_first=True, dtype=float)` để tạo ma trận nhị phân chuẩn số thực cho PyTorch.
   - Chuẩn hóa nhãn `SalePrice` qua hàm logarit `np.log1p` để phục vụ tối ưu hàm mất mát RMSLE.
  

````python
def preprocess_housing(train_df, test_df):
    n_train = len(train_df)
    
    # 1. Tách nhãn và gộp tập train (bỏ SalePrice) và test để đồng bộ không gian đặc trưng
    train_features = train_df.drop("SalePrice", axis=1)
    all_data = pd.concat([train_features, test_df], axis=0).reset_index(drop=True)
    
    # Bỏ cột Id vì không mang giá trị quy luật
    if 'Id' in all_data.columns:
        all_data = all_data.drop(columns=['Id'])
        
    # Chuyển các cột số mang bản chất phân loại thời vụ/kiến trúc sang chuỗi
    for cat_num_col in ['MSSubClass', 'MoSold']:
        if cat_num_col in all_data.columns:
            all_data[cat_num_col] = all_data[cat_num_col].astype(str)
    
    # 2. Xử lý các cột phân loại mà NA mang nghĩa "Không có tiện ích" -> Điền chuỗi 'None'
    none_cols = [
        'PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu',
        'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
        'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2',
        'MasVnrType'
    ]
    for col in none_cols:
        if col in all_data.columns:
            all_data[col] = all_data[col].fillna('None')
            
    # 3. Xử lý các cột số liên quan đến tầng hầm & nhà xe (Không có tiện ích -> Điền 0)
    zero_cols = [
        'GarageYrBlt', 'GarageArea', 'GarageCars',
        'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF',
        'BsmtFullBath', 'BsmtHalfBath', 'MasVnrArea'
    ]
    for col in zero_cols:
        if col in all_data.columns:
            all_data[col] = all_data[col].fillna(0)
            
    # 4. Điền LotFrontage theo trung vị của từng khu vực (Neighborhood)
    if 'LotFrontage' in all_data.columns:
        all_data['LotFrontage'] = all_data.groupby('Neighborhood')['LotFrontage'].transform(
            lambda x: x.fillna(x.median())
        )
        all_data['LotFrontage'] = all_data['LotFrontage'].fillna(all_data['LotFrontage'].median())
        
    # 5. Điền thiếu ngẫu nhiên cho các cột phân loại còn lại (MSZoning, Utilities, KitchenQual, Electrical...)
    cat_remaining = all_data.select_dtypes(include=['object', 'category', 'string']).columns
    for col in cat_remaining:
        if all_data[col].isnull().sum() > 0:
            all_data[col] = all_data[col].fillna(all_data[col].mode()[0])
            
    # 6. Điền thiếu ngẫu nhiên cho các cột số còn lại nếu có
    num_remaining = all_data.select_dtypes(include=['int64', 'float64']).columns
    for col in num_remaining:
        if all_data[col].isnull().sum() > 0:
            all_data[col] = all_data[col].fillna(all_data[col].median())
            
    # 7. One-Hot Encoding (ép dtype=float để tương thích tuyệt đối với StandardScaler và PyTorch)
    all_data = pd.get_dummies(all_data, drop_first=True, dtype=float)
    
    # 8. Tách lại train và test
    X = all_data.iloc[:n_train, :].copy()
    X_test_kaggle = all_data.iloc[n_train:, :].copy()
    
    # Biến đổi log1p cho biến mục tiêu SalePrice
    y = np.log1p(train_df["SalePrice"]).copy()
    
    return X, y, X_test_kaggle

# Chạy tiền xử lý
X, y, X_test_kaggle = preprocess_housing(train, test)

print("=" * 60)
print("KẾT QUẢ TIỀN XỬ LÝ")
print("=" * 60)
print(f"Shape X (Train)          : {X.shape}")
print(f"Shape y (Target Train)   : {y.shape}")
print(f"Shape X_test_kaggle (Test): {X_test_kaggle.shape}")
print(f"Đồng bộ số lượng cột     : {X.shape[1] == X_test_kaggle.shape[1]}")
print(f"Missing values còn lại   : {X.isnull().sum().sum()}")
print("=" * 60)
````

### 5.1. Phân chia tập dữ liệu, Chuẩn hóa & Tạo PyTorch Tensors
- Phân chia `X`, `y` thành tập huấn luyện (`X_train`, `y_train`) và tập kiểm định (`X_val`, `y_val`) với tỷ lệ validation lấy từ `params["data"]["validation_size"]`.
- Chuẩn hóa đặc trưng bằng `StandardScaler`: chỉ `fit_transform` trên `X_train`, sau đó áp dụng `transform` cho `X_val` và `X_test_kaggle` để tránh rò rỉ dữ liệu 
- Chuyển đổi dữ liệu sang dạng `torch.Tensor` kiểu `torch.float32`, trong đó nhãn mục tiêu `y` được định hình thành tensor cột `(N, 1)` bằng `.view(-1, 1).
- Đóng gói tập train thành `train_loader` với `TensorDataset` và `DataLoader` để nạp dữ liệu theo mini-batch.

````python
# 1. Phân chia tập Train và Validation

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=params["data"]["validation_size"],
    random_state=params["data"]["random_state"]
)

# 2. Chuẩn hóa đặc trưng (StandardScaler)
scaler = StandardScaler()

# Chỉ fit trên X_train để tránh rò rỉ dữ liệu
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_kaggle_scaled = scaler.transform(X_test_kaggle)

print("Kích thước sau khi chia tập:")
print(f"- X_train: {X_train.shape}, y_train: {y_train.shape}")
print(f"- X_val  : {X_val.shape}, y_val  : {y_val.shape}")
````

````python
# 3. Chuyển đổi sang PyTorch Tensor (float32)
X_train_t = torch.tensor(X_train_scaled, dtype=torch.float32)
y_train_t = torch.tensor(y_train.values, dtype=torch.float32).view(-1, 1)

X_val_t = torch.tensor(X_val_scaled, dtype=torch.float32)
y_val_t = torch.tensor(y_val.values, dtype=torch.float32).view(-1, 1)

X_kaggle_t = torch.tensor(X_test_kaggle_scaled, dtype=torch.float32)

# 4. Tạo DataLoader cho quá trình huấn luyện
train_loader = DataLoader(
    TensorDataset(X_train_t, y_train_t),
    batch_size=params["train"]["batch_size"],
    shuffle=True
)

# 5. Lưu và in thông tin bàn giao cho Member 2 & Member 3
input_size = X_train_t.shape[1]

print("=" * 70)

print("=" * 70)
print(f"- Kích thước Tensor X_train_t  : {X_train_t.shape}")
print(f"- Kích thước Tensor y_train_t  : {y_train_t.shape}")
print(f"- Kích thước Tensor X_val_t    : {X_val_t.shape}")
print(f"- Kích thước Tensor y_val_t    : {y_val_t.shape}")
print(f"- Kích thước Tensor X_kaggle_t : {X_kaggle_t.shape}")
print(f"- Số lượng batch trong train_loader : {len(train_loader)}")
print("-" * 70)
print(f">>> THÔNG SỐ INPUT_SIZE CHO MODEL PYTORCH: {input_size} <<<")
print("-" * 70)
print(f"Danh sách {len(X.columns)} đặc trưng đầu vào (Features):")
print(X.columns.tolist())
print("=" * 70)
````

## 6. Xây dựng Mô hình
Mô hình sử dụng các lớp Linear kết hợp hàm kích hoạt ReLU. Dropout và weight decay được sử dụng như các cơ chế regularization nhằm hạn chế overfitting. Kích thước các hidden layer được điều chỉnh thông qua quá trình tuning.

````python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# ================= PHẦN 6: ĐỊNH NGHĨA MÔ HÌNH =================
class HousePriceMLP(nn.Module):
    def __init__(self, input_size, hidden1=64, hidden2=32):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_size, hidden1),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(hidden1, hidden2),
            nn.ReLU(),
            nn.Linear(hidden2, 1)
        )

    def forward(self, x):
        return self.net(x)

# Khởi tạo mô hình với input_size từ bước trước
model = HousePriceMLP(
    input_size=input_size,
    hidden1=params["train"]["hidden1"],
    hidden2=params["train"]["hidden2"]
)
print(model)
````

## 7. Huấn luyện, Fit Mô hình

````python
# ================= PHẦN 7: HUẤN LUYỆN =================
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=params["train"]["learning_rate"],
    weight_decay=params["train"]["weight_decay"]
)

epochs = params["train"]["epochs"]
train_losses = []
val_losses = []

best_val_loss = float("inf")

print("--- BẮT ĐẦU HUẤN LUYỆN ---")
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    
    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        predictions = model(X_batch)
        loss = criterion(predictions, y_batch)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
        
    avg_train_loss = running_loss / len(train_loader)
    train_losses.append(avg_train_loss)
    
    model.eval()
    with torch.no_grad():
        val_preds = model(X_val_t)
        val_loss = criterion(val_preds, y_val_t).item()
        val_losses.append(val_loss)

    if val_loss < best_val_loss:
        best_val_loss = val_loss

        torch.save(
            model.state_dict(),
            best_model_path
        )

        print(
            f"  -> Lưu model tốt nhất tại epoch {epoch + 1}, "
            f"Val Loss = {val_loss:.6f}"
        )  
    
    if (epoch + 1) % 15 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {avg_train_loss:.4f}, Val Loss: {val_loss:.4f}')
````

## 8. Đánh giá Mô hình
Sử dụng kỹ thuật đảo ngược logarit (`np.expm1`) để chuyển giá trị đánh giá về lại đơn vị tiền tệ thực tế (Đô-la).

````python
# ================= PHẦN 8: ĐÁNH GIÁ MÔ HÌNH =================
model.load_state_dict(
    torch.load(best_model_path)
)

model.eval()
with torch.no_grad():
    # 1. Lấy giá trị logarit gốc do mô hình xuất ra
    y_val_pred_log = model(X_val_t).numpy()
    y_val_true_log = y_val_t.numpy()
    
    # 2. Sử dụng expm1 để đưa về giá tiền thật (chỉ dùng để xem MAE)
    y_val_pred_dollar = np.expm1(y_val_pred_log)
    y_val_true_dollar = np.expm1(y_val_true_log)

# Tính RMSE và R2 trên thang đo Logarit
mse = mean_squared_error(y_val_true_log, y_val_pred_log)
rmse = np.sqrt(mse)
r2 = r2_score(y_val_true_log, y_val_pred_log)

# Tính MAE trên tiền thực tế
mae = mean_absolute_error(y_val_true_dollar, y_val_pred_dollar)

print("--- ĐÁNH GIÁ MÔ HÌNH TRÊN TẬP VALIDATION ---")
print(f"Validation RMSE (log1p): {rmse:.4f}")
print(f"Validation MAE (USD):    ${mae:,.2f}")
print(f"Validation R2 (log1p):   {r2:.4f}")

# Vẽ đồ thị Loss
plt.figure(figsize=(10, 5))
plt.plot(train_losses, label='Train Loss')
plt.plot(val_losses, label='Validation Loss')
plt.title('Đồ thị hàm mất mát (MSE) qua các Epoch')
plt.xlabel('Epochs')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.show()
````

# PHẦN 9: TUNING

````python
# ================= 9. Tuning =================

def run_experiment(config):

    # 1. Tạo DataLoader theo batch_size của cấu hình
    loader = DataLoader(
        TensorDataset(X_train_t, y_train_t),
        batch_size=config["batch_size"],
        shuffle=True
    )

    # 2. Tạo model theo hidden layer của cấu hình
    model = HousePriceMLP(
        input_size=input_size,
        hidden1=config["hidden1"],
        hidden2=config["hidden2"]
    )

    # 3. Loss và Optimizer
    criterion = nn.MSELoss()

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=config["learning_rate"],
        weight_decay=config["weight_decay"]
    )

    # 4. Lưu validation loss tốt nhất
    best_val_loss = float("inf")
    best_state = None

    # 5. Huấn luyện
    for epoch in range(config["epochs"]):

        model.train()

        for X_batch, y_batch in loader:

            optimizer.zero_grad()

            predictions = model(X_batch)

            loss = criterion(predictions, y_batch)

            loss.backward()

            optimizer.step()

        # Validation
        model.eval()

        with torch.no_grad():
            val_preds = model(X_val_t)

            val_loss = criterion(
                val_preds,
                y_val_t
            ).item()

        if val_loss < best_val_loss:
            best_val_loss = val_loss

            best_state = {
                key: value.detach().cpu().clone()
                for key, value in model.state_dict().items()
            }

    # 6. Đổi MSE thành RMSE
    model.load_state_dict(best_state)

    val_rmse = np.sqrt(best_val_loss)

    return val_rmse, model
````
