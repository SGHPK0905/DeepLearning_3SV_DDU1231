# Check whether GPU (CUDA) can be used or not


```python
cuda_ok = torch.cuda.is_available()
print('CUDA available:', cuda_ok)
if cuda_ok:
    for i in range(torch.cuda.device_count()):
        print(f'GPU {i}:', torch.cuda.get_device_name(i))
else:
    print('Running on CPU:', torch.get_num_threads(), 'threads')
```

## Basic Tensor & Autograd verification


```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
x = torch.tensor([[1.0, 2.0], [3.0, 4.0]], requires_grad=True, device=device)
y = (2 * x ** 2 + 3).mean()
y.backward()
print('Active device:', device)
print('Gradient dy/dx:\n', x.grad)
```
